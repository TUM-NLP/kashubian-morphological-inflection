```python
from random import sample
%cd "/dss/dsshome1/03/ge87wod2/morphological-inflection/preprocessing/preprocessing_to_json_or_tsv"
!pip install torch transformers numpy pyarrow scikit-learn
import math, random, torch, json
from transformers import AutoTokenizer
tokenizer = AutoTokenizer.from_pretrained("google/byt5-small")
!tar -xzvf /dss/dsshome1/03/ge87wod2/morphological-inflection/data/leipzig-corpera-collection/pol_wikipedia_2021_10K.tar.gz
!tar -xzvf /dss/dsshome1/03/ge87wod2/morphological-inflection/data/leipzig-corpera-collection/csb_wikipedia_2021_10K.tar.gz 
pol_path = "pol_wikipedia_2021_10K/pol_wikipedia_2021_10K-sentences.txt"
csb_path = "csb_wikipedia_2021_10K/csb_wikipedia_2021_10K-sentences.txt"
random.seed(10)
with open(pol_path,"r") as pol_file, open (csb_path,"r") as csb_file:
    pol_lines = pol_file.readlines()[1:]
    csb_lines = csb_file.readlines()[1:]
    csb_lines =  sample(csb_lines,5000) 
    pol_lines =  sample(pol_lines,5000)
    csb_train, csb_dev = train_test_split (csb_lines,test_size = 0.1)
    pol_train, pol_dev = train_test_split (pol_lines,test_size = 0.1)
    import numpy as np
    from sklearn.model_selection import train_test_split
    datasets = ["trn", "dev"]
    split_data = {"trn": csb_train + pol_train,"dev": csb_dev + pol_dev}
    print(len(split_data["trn"]),len(split_data["dev"]))
    for dataset in datasets:
        out_path = "data/" + "csb-pol-mask-sent_"+dataset+".tsv"
        with open (out_path,"w") as out_file:
            for line in split_data[dataset]:
                sentence = line.strip().split("\t")[1] # cut out line number
                words = sentence.split(" ")
                size = 4
                end_index=len(words)-size
                while(end_index < 1): 
                    size-=1
                    end_index = len(words)-size
                mask_index = random.randint(1,end_index) # indices: [3 ... 23] until [-24 ... -4]
                masked_input = (  words[:mask_index] + ["<extra_id_0>"] + words[min (mask_index + 3,len(words)-1):]  )
                masked_labels = (["<extra_id_0>"]+ words[mask_index: min (mask_index + 3,len(words)-1)] +["<extra_id_1>"])
                x = " ".join(masked_input)+"\t"+" ".join(masked_labels) + "\n"
                out_file.write(x)
            print(out_path)
!rm -rf pol_wikipedia_2021_10K/
!rm -rf csb_wikipedia_2021_10K/
```
