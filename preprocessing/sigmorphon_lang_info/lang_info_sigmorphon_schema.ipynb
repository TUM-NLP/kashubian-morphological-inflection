```python
import math
sigmorphon_path = "<path_to>/kashubian-morphological-inflection/data/2023InflectionST/part1/data/"
slavic_languages = ["pol","ces","slk","dsb","bel","rus","mkd","csb"]
datasets = [".tst",".dev",".trn"]
target_lang = "pol-ces-slk-dsb-bel-rus-mkd-csb-family"
slavic_lang_code = "sla"
for dataset in datasets:
    out_path = sigmorphon_path + target_lang + dataset
    with open (out_path,"w") as out_file:
        for slavic_lang in slavic_languages:
            if slavic_lang == "csb" and dataset!=".tst": continue
            in_path = sigmorphon_path + slavic_lang + dataset
            with open (in_path,"r") as in_file:
                lines = in_file.readlines()
                size = len(lines)
                family_lines = [line.strip().split("\t")[0]+"\t"+line.strip().split("\t")[1]+";"+slavic_lang_code+"\t"+line.strip().split("\t")[2] + "\n" for line in lines]
                lang_lines = [line.strip().split("\t")[0]+"\t"+line.strip().split("\t")[1]+";"+slavic_lang+"\t"+line.strip().split("\t")[2]  + "\n" for line in lines]
                write_lines = lang_lines[:math.floor(size/2)] + family_lines[math.floor(size/2):]
                write_lines = family_lines
                join_lines = "".join(write_lines)
                out_file.write(join_lines)
                
```
